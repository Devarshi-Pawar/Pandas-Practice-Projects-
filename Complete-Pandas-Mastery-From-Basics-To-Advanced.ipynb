```python
employees
```

```python
salary
```

```python
pd.merge(employees,salary,on='emp_id',how='inner') #inner simply merge the data on the base of same rows or data which present in emp_id column
```

```python
pd.merge(employees,salary,on='emp_id',how='outer') #outer simply merge the table on the base of same column name 
```

```python
pd.merge(employees,salary,on='emp_id',how='left')
```

# Pivot Tables

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product')
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='count')
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='median')
```

```python
pivot2=pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product')
pivot2 #for aggregation function we use pivot table 
```

```python
df=pd.read_csv('/Users/devarshiabhaypawar/Desktop/data set/anime.csv')
```

```python
#making new column for episodes count
def extract_episodes(txt):
    check=False
    data=""

    for i in txt:
        if i==')':
            check=False
            return data
        if check==True:
            data=data+i
        if i=='(':
            check=True
            
```

```python
df["Episodes"]=df['Title'].apply(extract_episodes)

df
```

```python
df["Episodes"]=df["Episodes"].str.replace(" eps","")

df
```

```python
df["Episodes"]=df["Episodes"].astype(int)

df
```

```python
#make a new column for time stamp
df.loc[6]['Title']
```

```python
def extract_time(txt):
    check=False
    data=""

    for i in range(len(txt)):
        if txt[i]==")":
            for j in range (i+1,i+20):
                data=data+txt[j]
                
            return data 
```

```python
df['Total time']=df["Title"].apply(extract_time)

df
```

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1  # +1 to include the starting month
    except:
        return None

df['Months'] = df['Total time'].apply(calculate_total_months)
```
